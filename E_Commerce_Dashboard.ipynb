```python
import pandas as pd

print("=== E-COMMERCE SALES & PROFIT DASHBOARD ===")
print("The following files are available:")
print("- ecommerce_sales_part1.csv")
print("- ecommerce_sales_part2.csv")
print("- ecommerce_sales_part3.csv\n")

# 1. Taking input of the file name
file_name = input("Enter the CSV file name: ").strip()

try:
    # 2. Loading the required dataset
    df = pd.read_csv(file_name)

    # 3. Asking about no. of rows and printing it
    n = int(input("How many rows do you want to see? "))
    print(f"\nHere are the first {n} rows:")
    print(df.head(n))

    # 4. Calculating KPI's from dataset columns ('Sales' and 'Profit')
    total_sales = df["Sales"].sum()
    total_profit = df["Profit"].sum()
    total_orders = len(df)
    profit_margin = (total_profit / total_sales) * 100
    top_category = df.groupby("Category")["Sales"].sum().idxmax()

    # 5. Print Dashboard Summary
    print("\n=========================================")
    print("       E-COMMERCE KPI SUMMARY            ")
    print("=========================================")
    print(f"Total Orders   : {total_orders}")
    print(f"Total Revenue  : INR {total_sales:,.0f}")
    print(f"Total Profit   : INR {total_profit:,.0f}")
    print(f"Profit Margin  : {profit_margin:.2f}%")
    print(f"Top Category   : {top_category}")
    print("=========================================")

except FileNotFoundError:
    print(f"\nError: '{file_name}' was not found. Make sure it is in the same folder.")
except ValueError:
    print("\nError: Please enter a valid number for the rows.")
```
